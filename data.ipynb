```python
def get_available_agents(call_date):

    agents = dim_agents[
        (dim_agents["Hire Date"] <= call_date)
        &
        (
            (dim_agents["Exit Date"].isna())
            |
            (dim_agents["Exit Date"] >= call_date)
        )
    ]

    return agents
```

```python
def active_agents_in_month(year, month):

    end_of_month = pd.Timestamp(year, month, 1) + pd.offsets.MonthEnd(0)

    agents = dim_agents[
        (dim_agents["Hire Date"] <= end_of_month)
        &
        (
            (dim_agents["Exit Date"].isna())
            |
            (dim_agents["Exit Date"] >= end_of_month)
        )
    ]

    return agents
```

```python
def calculate_weight(agent_row, call_date):
    """
    حساب وزن الموظف حسب خبرته وقت المكالمة
    """

    experience_days = (call_date - agent_row["Hire Date"]).days

    if experience_days < 90:
        return 1.0

    elif experience_days < 180:
        return 1.2

    elif experience_days < 365:
        return 1.5

    else:
        return 2.0
```

```python
def choose_agent(call_date):

    available = get_available_agents(call_date).copy()

    available["Weight"] = available.apply(
        lambda row: calculate_weight(row, call_date),
        axis=1
    )

    selected = random.choices(
        population=available.to_dict("records"),
        weights=available["Weight"],
        k=1
    )[0]

    return selected
```

```python
def monthly_active_agents(year):
    """
    ترجع عدد الموظفين النشطين في نهاية كل شهر
    """

    data = []

    for month in range(1, 13):

        agents = active_agents_in_month(year, month)

        data.append({
            "Year": year,
            "Month": month,
            "Active Agents": len(agents)
        })

    return pd.DataFrame(data)
```

```python
def expand_month(month, target_calls):
    """
    تكبير عدد مكالمات شهر معين
    """

    template = get_month_template(month)

    expanded = template.sample(
        n=target_calls,
        replace=True,
        random_state=42
    ).reset_index(drop=True)

    return expanded
```

```python
def change_date(date_value, new_year, new_month):
    """
    تغيير السنة والشهر مع الحفاظ على اليوم إذا أمكن.
    """

    day = min(date_value.day, 28)

    return pd.Timestamp(
        year=new_year,
        month=new_month,
        day=day
    )
```

```python
def assign_agent(call_date):

    # الموظفين المتاحين وقت المكالمة
    available = dim_agents[
        (dim_agents["Hire Date"] <= call_date)
        &
        (
            (dim_agents["Exit Date"].isna())
            |
            (dim_agents["Exit Date"] >= call_date)
        )
    ]

    # اختيار Agent عشوائي
    selected = available.sample(1).iloc[0]

    return selected
```

```python
def assign_agents(df):

    df = df.copy()

    agent_ids = []
    agent_names = []
    teams = []
    leaders = []

    for call_date in df["Date"]:

        agent = assign_agent(call_date)

        agent_ids.append(agent["AgentID"])
        agent_names.append(agent["Agent"])
        teams.append(agent["Team"])
        leaders.append(agent["Team Leader"])

    df["AgentID"] = agent_ids
    df["Agent"] = agent_names
    df["Team"] = teams
    df["Team Leader"] = leaders

    return df
```

```python
from collections import defaultdict
import random

agent_call_counter = defaultdict(int)

def assign_agent(call_date):

    available = dim_agents[
        (dim_agents["Hire Date"] <= call_date)
        &
        (
            (dim_agents["Exit Date"].isna())
            |
            (dim_agents["Exit Date"] >= call_date)
        )
    ].copy()

    # نحسب Weight عكسي حسب عدد المكالمات الحالية
    weights = []

    for agent_id in available["AgentID"]:

        calls = agent_call_counter[agent_id]

        # كلما زاد عدد المكالمات يقل احتمال الاختيار
        weights.append(1 / (calls + 1))

    selected = random.choices(
        population=available.to_dict("records"),
        weights=weights,
        k=1
    )[0]

    agent_call_counter[selected["AgentID"]] += 1

    return selected
```

```python
def generate_year(year, target_calls):

    # توزيع المكالمات على الشهور
    month_targets = monthly_targets(target_calls)

    all_months = []

    for month in range(1, 13):

        # Template من نفس الشهر في 2021
        template = fact_2021_full[
            fact_2021_full["Date"].dt.month == month
        ].copy()

        # عدد المكالمات المطلوبة
        n_calls = month_targets.loc[month]

        # Sampling
        month_df = (
            template
            .sample(
                n=n_calls,
                replace=True,
                random_state=year * 100 + month
            )
            .reset_index(drop=True)
        )

        # تغيير السنة
        month_df["Date"] = month_df["Date"].apply(
            lambda d: d.replace(year=year)
        )

        # إعادة توزيع الـ Agents
        agent_call_counter.clear()

        month_df = assign_agents(month_df)

        all_months.append(month_df)

    # دمج السنة
    year_df = (
        pd.concat(all_months, ignore_index=True)
        .sort_values(["Date", "Time"])
        .reset_index(drop=True)
    )

    # Call ID
    year_df["Call Id"] = [
        f"{year}-{i:06d}"
        for i in range(1, len(year_df) + 1)
    ]

    return year_df
```

```python
def monthly_targets(total_calls):
    """
    توزيع عدد المكالمات على الشهور
    بناءً على توزيع 2021
    """

    monthly = (
        fact_2021_full["Date"]
        .dt.month
        .value_counts(normalize=True)
        .sort_index()
    )

    monthly = (monthly * total_calls).round().astype(int)

    # تصحيح فرق التقريب
    diff = total_calls - monthly.sum()
    monthly.iloc[-1] += diff

    return monthly
```

```python
import numpy as np

def generate_hold_time(answered):

    if answered == "N":
        return 0

    # 70% من المكالمات بدون Hold
    if np.random.rand() < 0.70:
        return 0

    # 30% فيها Hold
    return np.random.randint(5, 91)
```

```python
import numpy as np

def generate_acw(answered):

    # لو المكالمة غير مجابة
    if answered == "N":
        return 0

    # توزيع طبيعي بمتوسط 30 ثانية
    acw = np.random.normal(loc=30, scale=10)

    # منع القيم الشاذة
    acw = max(10, min(90, round(acw)))

    return acw
```

```python
def generate_sla_status(answered, speed):

    # المكالمة غير المجابة تعتبر خارج الـ SLA
    if answered == "N":
        return "Breached"

    if pd.isna(speed):
        return "Breached"

    if speed <= SLA_TARGET:
        return "Within SLA"

    return "Breached"
```

```python
import numpy as np

fcr_probability = {
    "Streaming": 0.90,
    "Admin Support": 0.85,
    "Payment related": 0.80,
    "Contract related": 0.75,
    "Technical Support": 0.60
}
```

```python
def generate_fcr(answered, resolved, topic):

    if answered == "N":
        return "No"

    if resolved == "N":
        return "No"

    probability = fcr_probability.get(topic, 0.75)

    return "Yes" if np.random.rand() < probability else "No"
```

```python
def add_fcr(df):

    df = df.copy()

    df["FCR"] = df.apply(
        lambda row: generate_fcr(
            row["Answered (Y/N)"],
            row["Resolved"],
            row["Topic"]
        ),
        axis=1
    )

    return df
```

```python
transfer_probability = {
    "Technical Support": 0.20,
    "Contract related": 0.12,
    "Payment related": 0.08,
    "Admin Support": 0.05,
    "Streaming": 0.03
}
```

```python
import numpy as np

def generate_transfer(topic, resolved, fcr):

    probability = transfer_probability.get(topic, 0.08)

    if fcr == "No":
        probability += 0.15

    if resolved == "N":
        probability += 0.20

    probability = min(probability, 0.90)

    return "Yes" if np.random.rand() < probability else "No"
```

```python
def add_transfer_flag(df):

    df = df.copy()

    df["Transfer Flag"] = df.apply(
        lambda row: generate_transfer(
            row["Topic"],
            row["Resolved"],
            row["FCR"]
        ),
        axis=1
    )

    return df
```

```python
import numpy as np

def generate_escalation(topic,
                        transfer,
                        resolved,
                        fcr,
                        satisfaction):

    probability = 0.02

    if transfer == "Yes":
        probability += 0.25

    if fcr == "No":
        probability += 0.20

    if resolved == "N":
        probability += 0.30

    if topic == "Technical Support":
        probability += 0.10

    if pd.notna(satisfaction):
        if satisfaction <= 2:
            probability += 0.25

    probability = min(probability, 0.95)

    return "Yes" if np.random.rand() < probability else "No"
```

```python
def add_escalation(df):

    df = df.copy()

    df["Escalation"] = df.apply(
        lambda row: generate_escalation(
            row["Topic"],
            row["Transfer Flag"],
            row["Resolved"],
            row["FCR"],
            row["Satisfaction rating"]
        ),
        axis=1
    )

    return df
```

```python
fact_2021_full = add_escalation(fact_2021_full)

fact_2022 = add_escalation(fact_2022)

fact_2023 = add_escalation(fact_2023)
```

```python
import numpy as np

def generate_login_hours(shift):

    if shift == "Morning":
        return round(np.random.uniform(8.0, 9.0), 2)

    elif shift == "Evening":
        return round(np.random.uniform(8.0, 9.0), 2)

    else:
        return round(np.random.uniform(7.5, 8.5), 2)
```
